```python
def total(*num):
    #print(num,type(num))
    return sum(num)

print(total(1,2,3))
print(total(1,2,3,4,5,6))
print(total())

```

```python
def ring_up(*prices):
    subtotal = sum(prices)
    tax = subtotal * 0.18                 # 18% sales tax
    return round(subtotal + tax,2)

print(ring_up(250))                       # a customer buying one item
print(ring_up(250, 1200, 80, 45))         # a customer with a full basket
```

```python
def create_order(customer,*items,**options):
    order={
        "customer":customer,
        "items":list(items),
        "shipping":"standard",
        "gift_card":False,
    }
    for key, value in order.items():
        order[key]=value
        return order
print(create_order("Ali","Phone"))
print(create_order("Fizza","PC",shipping="express",gift_card=True))
```
