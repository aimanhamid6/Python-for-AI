```python
class Student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def average(self):
        return sum(self.marks)/len(self.marks)

    def grade(self):
        avg=self.average()
        if avg>=80:
            return "A"
        elif avg>=70:
            return "B"
        elif avg>=60:
            return"C"
        else:
            return"F"
s=Student("Fatima",[85,92,78])
print(s.name)
print(s.average())
print(s.grade())
```
