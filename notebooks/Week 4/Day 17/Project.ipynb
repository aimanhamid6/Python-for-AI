```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

try:
    df = pd.read_csv("tips.csv")
except FileNotFoundError:
    sns.load_dataset("tips").to_csv("tips.csv", index=False) 
    df = pd.read_csv("tips.csv")
```

```python
df["gender"]=np.where(df["sex"]=="Female",1,0)
df["non_smoker"]=np.where(df["smoker"]=="Yes",1,0)
#df[["total_bill","tip","sex","smoker","gender","non_smoker"]].head()
```

```python
print(round(df["tip"].mean(),1))
```

```python
df.groupby("sex")["tip"].mean()
```

```python
sns.boxplot(x="sex", y="tip", data=df)

plt.title("Tips by Gender")
plt.xlabel("Gender")
plt.ylabel("Tip")

plt.show()
```

```python
sns.boxplot(x="smoker", y="tip", data=df)

plt.title("Tips by Gender")
plt.xlabel("Smoker or Non- smoker")
plt.ylabel("Tip")

plt.show()
```
